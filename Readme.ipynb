# Customer Feedback Analysis & Automated Apology Email Generator

## Project Overview

This project analyzes customer reviews from an e-commerce dataset to identify critical customer feedback and automatically generate personalized apology emails using Generative AI. The objective is to demonstrate data cleaning, rule-based text analysis, and Large Language Model (LLM) integration using Python.

---

## Objectives

* Load and preprocess customer review data.
* Handle missing values in the dataset.
* Clean review text by removing special characters and converting text to lowercase.
* Identify critical reviews based on customer ratings.
* Extract the most common complaint keywords using rule-based logic.
* Generate personalized apology emails for the most detailed critical reviews using Generative AI.

## Dataset

The project uses an e-commerce customer review dataset (`flipkart_product.csv`) containing customer reviews and ratings.

Key columns used:

* **Review** – Customer review text
* **Rate** – Customer rating (1–5)

---

## Data Cleaning Approach

The following preprocessing steps are performed before analysis:

1. Load the CSV dataset using Pandas.
2. Check for missing values.
3. Remove rows where the **Review** column is missing.
4. Convert all review text to lowercase.
5. Remove punctuation, numbers, and special characters using Regular Expressions.
6. Remove leading and trailing whitespace.

These steps ensure consistent and clean text for keyword analysis.

---

## Rule-Based Logic

The project uses rule-based filtering instead of machine learning.

### Critical Review Identification

* Convert the **Rate** column to numeric.
* Reviews with **Rating ≤ 2** are classified as **Critical Reviews**.

### Complaint Keyword Extraction

* Split each review into individual words.
* Remove common stop words (such as *the, is, and, of,* etc.).
* Count the frequency of remaining words using Python's `collections.Counter`.
* Display the most common complaint keywords.

### Review Selection

To generate meaningful AI responses, the project selects the **three longest critical reviews**, as they typically contain more detailed customer feedback.

---

## Generative AI Integration

The project uses:

* **LangChain**
* **Groq API**
* **Llama 3.3 70B Versatile**

The AI model acts as a professional customer support representative and generates personalized apology emails based on each customer's review.

Each email:

* Apologizes sincerely.
* Addresses the customer's concerns.
* Assures the customer that the issue will be investigated.
* Offers further assistance.
* Maintains a professional and empathetic tone.
